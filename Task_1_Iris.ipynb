# Task 1: Exploring and Visualizing a Simple Dataset

## Introduction

The Iris dataset contains measurements of iris flowers.

## Problem Statement

The objective is to explore the dataset and visualize the data.

## Objectives

- Load the dataset using pandas.
- Understand its structure.
- Create scatter plot, histogram, and box plot.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

print("Libraries imported successfully!")
```

```python
iris = load_iris()

df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = pd.Categorical.from_codes(
    iris.target,
    iris.target_names
)

print("Dataset loaded successfully!")
df.head()
```

```python
print("Number of rows and columns:")
print(df.shape)
```

```python
print(df.columns)
```

```python
df.head(10)
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Total duplicate rows:")
print(df.duplicated().sum())
```

```python
df = df.dropna().copy()
df = df.drop_duplicates().reset_index(drop=True)

print("Data cleaning completed!")
print("New dataset shape:", df.shape)
print("Missing values remaining:", df.isnull().sum().sum())
print("Duplicate rows remaining:", df.duplicated().sum())
```

```python
print(df["species"].value_counts())
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="sepal length (cm)",
    y="petal length (cm)",
    hue="species",
    s=70
)

plt.title("Sepal Length vs Petal Length")
plt.xlabel("Sepal Length (cm)")
plt.ylabel("Petal Length (cm)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    data=df,
    x="petal length (cm)",
    bins=15,
    kde=True
)

plt.title("Distribution of Petal Length")
plt.xlabel("Petal Length (cm)")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="species",
    y="petal length (cm)"
)

plt.title("Petal Length by Species")
plt.xlabel("Species")
plt.ylabel("Petal Length (cm)")
plt.tight_layout()
plt.show()
```

```python
numeric_df = df.select_dtypes(include="number")
correlation = numeric_df.corr()

plt.figure(figsize=(8, 5))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Between Iris Features")
plt.tight_layout()
plt.show()
```

## Conclusion

The Iris dataset was loaded and explored using pandas.

The dataset structure, column names, statistical summary,
missing values, and duplicate rows were examined.

A scatter plot was used to explore the relationship between
